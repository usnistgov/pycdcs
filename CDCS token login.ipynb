# CDCS token login

This Notebook provides additional information regarding the recommended token login method.

## 1. Get a token

The first step is that you need to get an API token for the CDCS instance you wish to log in to.

- If you do not have admin privileges to the database, email the admin and ask them to create a token for you.
- If you do have admin privleges, you can create a new token for yourself or others by logging in to the admin dashboard, and going to the "Tokens" page listed under "AUTH TOKEN".

Once you have a token, save it somewhere secure.  The two recommended locations are in a file in your user directory, or by setting it as an environmental variable.

## 2. Logging in

The CDCS.use_token() method will create a CDCS interface using only the host url and the token.  The token can be specified in a variety of formats

- Give the token directly as a string.  *This is not recommended as you can accidentally reveal your token to others!*
- Give a path to a file containing the token.
- Give a name of an environmental variable, which can be the token or a file containing the token.
- If token=None or is not given, a secure prompt will ask for the token which can be given in any of the above formats.

```python
import os
from cdcs import CDCS
```

```python
# Call use_token() without giving the token.  A prompt will then ask for it.
curator = CDCS.use_token(host='https://potentials.nist.gov/')
print(curator.cdcsversion)
```

```python
# Call use_token() and give a path to a file containing the token.
token='C:/Users/lmh1/tokens/potentials.nist.gov.txt'
curator = CDCS.use_token(host='https://potentials.nist.gov/', token=token)
print(curator.cdcsversion)
```

```python
# Temporarily set an environmental variable for demonstration purposes
os.environ['POTENTIALS_API_KEY'] = token

# Call use_token() and give the environmental variable name
curator = CDCS.use_token(host='https://potentials.nist.gov/', token='POTENTIALS_API_KEY')
print(curator.cdcsversion)
```
